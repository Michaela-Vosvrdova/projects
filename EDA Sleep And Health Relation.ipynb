## **Exploratory Data Analysis**

```python
sns.histplot(df['Age'], kde=True, bins=20, color='red')
plt.title('Age Distribution')
sns.despine()
plt.show()
```

```python
sns.countplot(data=df, x='Gender', color='green')
plt.title('Gender Counts')
sns.despine()
plt.show()
```

```python
sns.boxplot(data=df, x='Gender', y='Age', color='purple')
plt.title('Age Distribution by Gender')
sns.despine()
plt.show()
```

```python
order = ['Normal', 'Normal Weight', 'Overweight', 'Obese']
sns.boxplot(data=df, x='BMI Category', y='Quality of Sleep', order=order, color='yellow')
plt.title('Quality of Sleep by BMI')
sns.despine()
plt.show()
```

```python
sns.violinplot(data=df, x='Gender', y='Sleep Duration', color='blue')
plt.title('Sleep Duration by Gender')
sns.despine()
plt.show()
```

```python
sns.kdeplot(data=df, x='Stress Level', y='Heart Rate', color='orange')
plt.title('Stress Level and Heart Rate Relation')
sns.despine()
plt.show()
```

```python
sns.countplot(df['Occupation'], color='green')
plt.title('Occupation Counts')
sns.despine()
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=df, x='Occupation', y='Stress Level', color='pink')
plt.title('Stress Level by Occupation')
plt.xticks(rotation=45)
plt.tight_layout()
sns.despine()
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=df, x='Occupation', y='Quality of Sleep', color='blue')
plt.title('Stress Level by Occupation')
plt.xticks(rotation=45)
plt.tight_layout()
sns.despine()
plt.show()
```

```python
sns.boxplot(data=df, x='Gender', y='Daily Steps', color='brown')
plt.title('Daily Steps by Gender')
sns.despine()
plt.show()
```

From the plots above we can draw the following conclusions:
- the more obese a man is, the worse his sleep quality tends to be,
- women sleep slightly longer than men,
- sales representatives, salespeople, and scientists experience more stress and have poorer sleep quality than people in other occupations,
- men take slightly more daily steps than women.

```python
sns.pairplot(df)
```

```python
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
```

From the heatmap above, we can draw the following conclusions based on the numerical values:

- the longer the sleep duration, the better the sleep quality,
- high stress levels negatively impact a man's heart rate, sleep quality, and sleep duration,
- younger people tend to experience slightly more stress than older individuals,
- surprisingly, physical activity does not seem to improve stress levels, and the number of daily steps appears to have no effect on sleep quality or sleep duration.
