```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
df = pd.read_csv('Student Mental health.csv')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
print('Number of duplicated rows: ', df.duplicated().sum())
print('Number of empty rows: ', df.isnull().sum())
```

```python
df = df.rename(columns={'Choose your gender': 'Gender', 'What is your course?': 'Course',
                       'Your current year of Study': 'Year of Study', 'What is your CGPA?': 'CGPA',
                       'Do you have Depression?': 'Depression', 'Do you have Anxiety?': 'Anxiety',
                       'Do you have Panic attack?': 'Panic Attack', 'Did you seek any specialist for a treatment?': 'Specialist Treatment'})
```

```python
df['Year of Study'] = df['Year of Study'].str.strip().str.capitalize()
```

```python
sns.histplot(data=df, x='Age', hue='Gender', multiple='stack', kde=False)
plt.title('Age and Gender Distribution')
plt.show()
```

```python
top_courses = df['Course'].value_counts().nlargest(5).index
filtered_df = df[df['Course'].isin(top_courses)]
order = top_courses

sns.countplot(data=filtered_df, x='Course', order=order, color='green')
plt.title('Course Counts')
plt.show()
```

```python
g = sns.catplot(data=df, x='Marital status', col='Year of Study', kind='count', color='orange')
plt.subplots_adjust(top=0.75)
plt.suptitle("Marital Status by Year of Study", fontsize=16)
plt.show()
```

```python
sns.countplot(data=df, x='Specialist Treatment', hue='Gender', palette='dark:pink')
plt.title('Specialist Treatment by Gender')
plt.show()
```

```python
sns.countplot(data=df, x='Gender', hue='Depression', palette='dark:yellow')
plt.title('Specialist Treatment by Gender')
plt.show()
```
